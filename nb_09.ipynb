```python

N = 10_000_000

A = rng.random(N) < 0.01
p_defect = np.where(A, 1, 0.05) 
defect = rng.random(N) < p_defect

print(A[defect].mean())
```

```python
omega = rng.choice(4, size=N, p=[0.25, 0.25, 0.25, 0.25]) 

A = np.isin(omega, [0, 1])
B = np.isin(omega, [1, 2])  
C = np.isin(omega, [3, 1]) 

print((A & B & C).mean())
print((A & B).mean() * (B & C).mean() * (A & C).mean())
```
